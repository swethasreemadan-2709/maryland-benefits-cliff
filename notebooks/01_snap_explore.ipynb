```python
import sys
sys.path.append("..")

from cliff.household import Household
from cliff.params import load_params
from cliff.snap import snap_benefit

params = load_params()

for earnings in range(0, 50_001, 5_000):
    family = Household(adults=1, child_ages=[3], annual_earnings=earnings,
                       rent_monthly=params["housing"]["rent_monthly"])
    print(f"${earnings:>7,}  ->  SNAP ${snap_benefit(family, params):>4} per month")
```

```python
from cliff.medicaid import medicaid_status
from cliff.childcare import childcare_costs

for earnings in [25_000, 30_000, 40_000, 50_000, 70_000, 76_000, 77_000]:
    family = Household(adults=1, child_ages=[3], annual_earnings=earnings,
                       rent_monthly=params["housing"]["rent_monthly"])
    health = medicaid_status(family, params)
    care = childcare_costs(family, params)
    print(f"${earnings:>6,} | parent covered: {health['adults_covered']} "
          f"| child: {health['children'][0]} "
          f"| child care paid: ${care['family_pays']:>9,.0f}")
```
