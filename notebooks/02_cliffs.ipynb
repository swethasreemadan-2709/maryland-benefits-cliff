```python
import matplotlib.pyplot as plt

results = pd.read_csv("../data/results.csv")
a = results[results["household"].str.startswith("A")]

for scenario, data in a.groupby("scenario"):
    plt.plot(data["earnings"], data["net_resources"], label=scenario)
plt.plot(a["earnings"], a["earnings"], "--", color="gray", label="earnings alone")
plt.xlabel("Annual earnings ($)")
plt.ylabel("Net resources ($)")
plt.title("Household A: single parent, one child")
plt.legend()
plt.show()
```
